# 📊 Indian E-Commerce Product Analytics using Array Operations and Algorithms

## 📌 Project Objectives:
1. Implement basic array operations: **Insertion, Deletion, Searching, Sorting, Merging**.
2. Use advanced array algorithms like **Sliding Window, Two Pointers, Kadane’s Algorithm, and Prefix Sum**.


```python
# Step 1: Initialize the Product Price List (Array) and Insert Prices
product_prices = []  # Empty array to store product prices (in INR)

def add_product(price):
    """Inserts a product price into the array."""
    product_prices.append(price)  # Inserting into array (O(1) operation)
    print(f'Product with price ₹{price} added!')

# Adding some initial prices (Example: Mobile phone prices in INR)
add_product(14999)
add_product(22999)
add_product(9999)
add_product(17999)
print("Current Product Prices:", product_prices)
```

```python
# Step 2: Remove a Product Price from the List
def remove_product(price):
    """Removes a product price if it exists in the array."""
    if price in product_prices:
        product_prices.remove(price)  # Deleting from array (O(n) operation)
        print(f'Product with price ₹{price} removed!')
    else:
        print(f'Product with price ₹{price} not found!')
remove_product(22999)
print("Updated Product Prices:", product_prices)
```

```python
# Step 3: Search for a Product Price (Linear Search)
def search_product(price):
    """Searches for a product price using linear search."""
    for i in range(len(product_prices)):  # O(n) complexity
        if product_prices[i] == price:
            print(f'Product with price ₹{price} found at index {i}')
            return
    print(f'Product with price ₹{price} not found!')
search_product(14999)
search_product(5000)
```

```python
# Step 4: Sort the Product Prices (Bubble Sort)

def bubble_sort_prices():
    """Sorts the product prices using Bubble Sort algorithm."""
    n = len(product_prices)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if product_prices[j] > product_prices[j + 1]:  # Swap if greater
                product_prices[j], product_prices[j + 1] = product_prices[j + 1], product_prices[j]
    print("Sorted Product Prices:", product_prices)

bubble_sort_prices()
```

```python
# Step 5: Merge Two Product Price Lists
more_products = [32999, 19999]  # Prices of new products

def merge_prices(list1, list2):
    """Merges two lists into one."""
    return list1 + list2  # Merging two arrays (O(m+n) complexity)

product_prices = merge_prices(product_prices, more_products)
print("Merged Product Prices:", product_prices)
```

```python
# Step 6: Sliding Window Algorithm - Find Top K Total Price

def max_sum_sliding_window(arr, k):
    """Finds maximum sum of K consecutive product prices using Sliding Window."""
    max_sum = sum(arr[:k])  # Initial window sum
    window_sum = max_sum
    for i in range(len(arr) - k):
        window_sum = window_sum - arr[i] + arr[i + k]  # Slide the window
        max_sum = max(max_sum, window_sum)
    return max_sum

print("Max total price of 2 consecutive products:", max_sum_sliding_window(product_prices, 2))
```

```python
# Step 7: Two Pointers - Find Two Products with Target Sum

def find_two_products_with_sum(arr, target):
    """Finds two product prices in a sorted array that add up to the target sum using Two Pointers."""
    arr.sort()
    left, right = 0, len(arr) - 1
    while left < right:
        curr_sum = arr[left] + arr[right]
        if curr_sum == target:
            return arr[left], arr[right]
        elif curr_sum < target:
            left += 1
        else:
            right -= 1
    return "No pair found"

print("Products with sum ₹40000:", find_two_products_with_sum(product_prices, 27998))
```

```python
# Step 8: Kadane’s Algorithm - Maximum Sum Subarray
def max_subarray_kadane(arr):
    """Finds the maximum sum of any contiguous subarray using Kadane’s Algorithm."""
    max_sum = curr_sum = arr[0]
    for num in arr[1:]:
        curr_sum = max(num, curr_sum + num)
        max_sum = max(max_sum, curr_sum)
    return max_sum

print("Maximum total price of a contiguous product subarray:", max_subarray_kadane(product_prices))
```

```python
# Step 9: Prefix Sum - Get Sum of Prices in Range

def prefix_sum(arr):
    """Computes prefix sum array for quick range sum queries."""
    prefix = [0] * (len(arr) + 1)
    for i in range(1, len(arr) + 1):
        prefix[i] = prefix[i - 1] + arr[i - 1]
    return prefix

prefix = prefix_sum(product_prices)

def range_sum(l, r):
    """Returns sum of elements between indices l and r."""
    return prefix[r + 1] - prefix[l]

print("Total price of products from index 1 to 3:", range_sum(1, 3))
```
